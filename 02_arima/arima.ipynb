# ARIMA Model

Fit an ARIMA model to a time series using `statsmodels`.

```python
# Install dependencies (if needed)
# !pip install statsmodels arch pandas numpy matplotlib
```

```python
import numpy as np
from statsmodels.tsa.arima.model import ARIMA

np.random.seed(42)
y = np.cumsum(np.random.randn(200))

model = ARIMA(y, order=(1, 1, 1))
result = model.fit()
print(result.summary())
```
