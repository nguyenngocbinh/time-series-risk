# ARCH Model

Fit an ARCH model to capture volatility clustering.

```python
# Install dependencies (if needed)
# !pip install statsmodels arch pandas numpy matplotlib
```

```python
import numpy as np
from arch import arch_model

np.random.seed(42)
returns = np.random.randn(500) * 0.01

am = arch_model(returns, vol='ARCH', p=1)
res = am.fit()
print(res.summary())
```
