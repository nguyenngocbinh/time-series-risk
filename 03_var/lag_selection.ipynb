# VAR Lag Selection

Select the optimal lag order for a VAR model using information criteria.

```python
# Install dependencies (if needed)
# !pip install statsmodels arch pandas numpy matplotlib
```

```python
import numpy as np
import pandas as pd
from statsmodels.tsa.api import VAR

np.random.seed(42)
data = pd.DataFrame(np.random.randn(200, 2), columns=['y1', 'y2'])

model = VAR(data)
result = model.select_order(maxlags=10)
print(result.summary())
```
