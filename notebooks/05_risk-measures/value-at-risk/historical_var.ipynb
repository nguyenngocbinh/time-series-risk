# Historical VaR (Value at Risk)

Compute Value at Risk using the historical simulation method.

```python
# Install dependencies (if needed)
# !pip install statsmodels arch pandas numpy matplotlib
```

```python
import numpy as np
import pandas as pd

np.random.seed(42)
returns = pd.Series(np.random.randn(1000) * 0.01)

confidence_level = 0.95
var_historical = np.percentile(returns, (1 - confidence_level) * 100)
print(f'Historical VaR ({confidence_level*100:.0f}%): {var_historical:.4f}')
```
