# Parametric VaR (Value at Risk)

Compute Value at Risk using the parametric (variance-covariance) method.

```python
# Install dependencies (if needed)
# !pip install statsmodels arch pandas numpy matplotlib
```

```python
import numpy as np
from scipy.stats import norm

np.random.seed(42)
returns = np.random.randn(1000) * 0.01

mu = np.mean(returns)
sigma = np.std(returns)
confidence_level = 0.95

var_parametric = mu + sigma * norm.ppf(1 - confidence_level)
print(f'Parametric VaR ({confidence_level*100:.0f}%): {var_parametric:.4f}')
```
