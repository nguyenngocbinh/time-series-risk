# Impulse Response Functions

Compute and plot impulse response functions from a fitted VAR model.

```python
# Install dependencies (if needed)
# !pip install statsmodels arch pandas numpy matplotlib
```

```python
import numpy as np
import pandas as pd
from statsmodels.tsa.api import VAR
import matplotlib.pyplot as plt

np.random.seed(42)
data = pd.DataFrame(np.random.randn(200, 2), columns=['y1', 'y2'])

model = VAR(data)
result = model.fit(2)
irf = result.irf(10)
irf.plot()
plt.show()
```
