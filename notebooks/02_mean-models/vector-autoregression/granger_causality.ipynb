# Granger Causality

Test for Granger causality between time series variables.

```python
# Install dependencies (if needed)
# !pip install statsmodels arch pandas numpy matplotlib
```

```python
import numpy as np
import pandas as pd
from statsmodels.tsa.api import VAR
from statsmodels.tsa.stattools import grangercausalitytests

np.random.seed(42)
data = pd.DataFrame(np.random.randn(200, 2), columns=['y1', 'y2'])

grangercausalitytests(data[['y1', 'y2']], maxlag=4)
```
