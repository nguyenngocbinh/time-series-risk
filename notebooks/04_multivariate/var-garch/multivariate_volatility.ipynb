# Multivariate Volatility (VAR-GARCH)

Combine VAR and GARCH models to capture multivariate volatility dynamics.

```python
# Install dependencies (if needed)
# !pip install statsmodels arch pandas numpy matplotlib
```

```python
import numpy as np
import pandas as pd
from arch import arch_model
from statsmodels.tsa.api import VAR

np.random.seed(42)
data = pd.DataFrame(np.random.randn(500, 2) * 0.01, columns=['asset1', 'asset2'])

# Step 1: Fit VAR to get residuals
var_model = VAR(data)
var_res = var_model.fit(1)
residuals = pd.DataFrame(var_res.resid, columns=['resid1', 'resid2'])

# Step 2: Fit GARCH on each residual series
for col in residuals.columns:
    am = arch_model(residuals[col], vol='Garch', p=1, q=1)
    res = am.fit(disp='off')
    print(f'--- {col} ---')
    print(res.summary())
```
