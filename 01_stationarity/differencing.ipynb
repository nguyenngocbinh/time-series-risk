# Differencing for Stationarity

Apply first and seasonal differencing to make a time series stationary.

```python
# Install dependencies (if needed)
# !pip install statsmodels arch pandas numpy matplotlib
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

np.random.seed(42)
y = np.cumsum(np.random.randn(200))

dy = pd.Series(y).diff().dropna()

fig, axes = plt.subplots(2, 1, figsize=(10, 6))
axes[0].plot(y, label='Original')
axes[0].legend()
axes[1].plot(dy, label='1st Difference')
axes[1].legend()
plt.tight_layout()
plt.show()
```
