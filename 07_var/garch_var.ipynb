# GARCH-based VaR (Value at Risk)

Compute Value at Risk using GARCH-forecasted conditional volatility.

```python
# Install dependencies (if needed)
# !pip install statsmodels arch pandas numpy matplotlib
```

```python
import numpy as np
from arch import arch_model
from scipy.stats import norm

np.random.seed(42)
returns = np.random.randn(500) * 0.01

am = arch_model(returns, vol='Garch', p=1, q=1)
res = am.fit(disp='off')
forecasts = res.forecast(horizon=1)
forecast_vol = np.sqrt(forecasts.variance.values[-1, 0])

confidence_level = 0.95
var_garch = norm.ppf(1 - confidence_level) * forecast_vol
print(f'GARCH VaR ({confidence_level*100:.0f}%): {var_garch:.4f}')
```
